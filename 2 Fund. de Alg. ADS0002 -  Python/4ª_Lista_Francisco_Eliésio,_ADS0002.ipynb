```python
#________________________________________________________________________________________________
# FUNÇÕES PARA VALIDAR OS INPUTS NAS QUESTÕES
#________________________________________________________________________________________________

#Função que verifica se a string do input pode ser convertida em int.
def ni(n):
  while True:
    try:
      n = int(n)
      return n
      break
    except (ValueError, TypeError):
      print('Você deve digitar um número inteiro!')
      n = input('Digite um número: ')
      return ni(n)

# Função que verifica se a string do input pode ser convertida em float.
def nf(n):
  while True:
    try:
      n = float(n)
      return n
      break
    except (ValueError, TypeError):
      print('Você deve digitar um número!')
      n = input('Digite um número: ')
      return nf(n)
```

```python
#________________________________________________________________________________________________
# Algoritmo: listaInversa
# Propósito: ler dez números e imprimi-los e ordem inversa numa lista.
# Retorno: lista inversa à ordem de entrada de dados.
#________________________________________________________________________________________________

lista = []
while len(lista) < 10:
  lista.append(nf(input(f'Digite o {len(lista)+1}° número da lista: ')))
lista.reverse()

mensagem = f'Lista na ordem inversa: {lista}'
print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**4.** Faça um Programa que leia uma lista de 10 caracteres, e diga quantas consoantes foram lidas. Imprima as consoantes.

```python
import re

utf8 = ''.join(chr(i) for i in range(32, 256))
let = re.findall('[b-df-hj-np-tv-z]', utf8)

caracteres = []
numero_de_catacteres = 10
while len(caracteres) < numero_de_catacteres:
  caracteres.append(input(f'Digite o {len(caracteres)+1}° caractere: '))

consoantes = set([letra for letra in caracteres if letra in let])
mensagem = f'Número de consoantes: {len(consoantes)}\n'f'Consoantes: {consoantes}'
print(mensagem)
```

**5.** Faça um Programa que leia 20 números inteiros e armazene-os numa lista. Armazene os números pares na lista PAR e os números IMPARES na lista impar. Imprima as três listas.

```python
numeros = []
while len(numeros) < 20:
  numeros.append(ni(input(f'Digite o {len(numeros)+1}° número: ')))

par = [numero for numero in numeros if numero % 2 == 0]
impar = [numero for numero in numeros if numero % 2 != 0]

print(numeros)
print(par)
print(impar)
```

**6.** Faça um Programa que peça as quatro notas de 10 alunos, calcule e armazene numa lista a média de cada aluno, imprima o número de alunos com média maior ou igual a 7.0.

```python
def verificar_nota():
    while True:
        try:
            nota = float(input('Digite uma nota válida: '))
            if 0 <= nota <= 10:
                return nota
            else:
                print("A nota deve estar entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número.")

notas = 2  # Número de notas por aluno
alunos = 2  # Número de alunos

medias = []

while len(medias) < alunos:
    print(f'Digite as notas do {len(medias)+1}º estudante:')
    notas_alunos = []
    while len(notas_alunos) < notas:
        nota = verificar_nota()  # Valida a nota antes de adicionar
        notas_alunos.append(nota)
    medias.append(sum(notas_alunos) / len(notas_alunos))  # Calcula a média

print("Médias dos alunos:", medias)
```

```python
#________________________________________________________________________________________________
# Algoritmo: mediaDezAlunos
# Propósito: ler quatro notas de dez alunos e imprimir a quantidade de médias maior ou igual a sete
# Retorno: número de alunos com média maior que sete
#________________________________________________________________________________________________

medias = []

for i in range(10):
  print(f'DIGITE AS NOTAS DO {i+1}° ESTUDANTE: ')
  notas = []
  for j in range(4):
    n = -1
    while n < 0 or n > 10:
      n = nf(input(f'Digite a {j+1}ª nota: '))
    notas.append(n)
  medias.append(round((sum(notas)/len(notas)),1))

contador = 0
for i in medias:
  if i >= 7:
    contador += 1

mensagem = f'Há {contador} estudante(s) com média igual ou maior que 7.0.\n'f'Médias: {medias}'
print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**7.** Faça um Programa que leia uma lista de 5 números inteiros, mostre a soma, a multiplicação e os números.

```python
#________________________________________________________________________________________________
# Algoritmo: aritmeticaLista
# Propósito: ler cinco números inteiros em uma lista, mostrá-los, e efetuar a soma e a multiplicação de tais números.
# Retorno: resultado da soma e da multiplicação de cinco números.
#________________________________________________________________________________________________
lista = []

for i in range(5):
  lista.append(ni(input(f'Digite o {i+1}° número inteiro: ')))

s = 0
m = 1
for i in lista:
  m = m*i
  s += i

mensagem = f'Lista de números: {lista}\n'f'Soma dos números: {s}\n'f'Multiplicação dos números: {m}'

print(f'{len(mensagem)//2*"-"}\n{mensagem}\n{len(mensagem)//2*"-"}')
```

**8.** Faça um Programa que peça a idade e a altura de 5 pessoas, armazene cada informação na sua respectiva lista. Imprima a idade e a altura na ordem inversa a ordem lida.

```python
#________________________________________________________________________________________________
# Algoritmo: idadeAlturaPessoa
# Propósito: receber a idade e a altura de cinco pessoas e imprimir as listas de dados em ordem inversa.
# Retorno: lista de idade e altura em ordem inversa
#________________________________________________________________________________________________

n = 5
lista = []
for i in range(n):
  lista.append([i+1,'idade','altura'])

for i in range(n):
  print(f'DIGITE OS DADOS DA {lista[i][0]}ª PESSOA: ')
  lista[i][1] = ni(input(f'Idade (em anos): '))
  lista[i][2] = nf(input(f'Altura (em metros): '))

lista.reverse()
mensagem = m = ''
for i in range(n):
  m = f'{lista[i][0]}ª pessoa: {lista[i][1]} anos e altura {lista[i][2]} m.\n'
  mensagem += m

print(f'{len(mensagem)//2*"-"}\n{mensagem}\n{len(mensagem)//2*"-"}')
```

**10.** Faça um Programa que leia duas listas com 10 elementos cada. Gere uma terceira lista de 20 elementos, cujos valores deverão ser compostos pelos elementos intercalados dos duas outras listas.

```python
#________________________________________________________________________________________________
# Algoritmo: intercalarDuasListas
# Propósito: receber duas listas de dez elementos e criar uma terceira lista com os elementos da primeira e segunda lista intercalados.
# Retorno: lista com elementos intercalados
#________________________________________________________________________________________________

numElemento = 10
numLista = 2
lista = []
for i in range(numLista):
  lista.append(list(range(numElemento)))

for i in range(numLista):
  print(f'DIGITE OS ELEMENTOS DA {i+1}ª LISTA: ')
  for j in range(numElemento):
    lista[i][j] = input(f'Digite o {j+1}° elemento: ')

novaLista = []
for i in range(numElemento):
  for j in range(numLista):
    novaLista.append(lista[j][i])

mensagem = ''
for i in range(numLista):
  w = f'{i+1}ª Lista: {lista[i]}\n'
  mensagem += w

mensagem = f'{mensagem}{numLista+1}ª Lista: {novaLista}'

print(f'{len(mensagem)//2*"-"}\n{mensagem}\n{len(mensagem)//2*"-"}')
```
