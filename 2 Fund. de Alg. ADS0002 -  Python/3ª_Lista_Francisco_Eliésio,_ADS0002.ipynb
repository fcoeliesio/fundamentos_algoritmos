**7.** Escreva um programa que exiba uma lista de opções (menu): Adição, subtração, divisão e multiplicação e sair. Imprima o resultado da operação escolhida entre dois números informados pelo usuário. Repita até que a opção sair seja escolhida.

```python
menu = [['Adição',1],['Subtração',2],['Divisão',3],['Multiplicação',4],['Sair',5]]
e = 37*"_"
print(f'{e} MENU {e}')
for i in range(int(len(menu))):
  print(f'| {menu[i][0]} ({menu[i][1]}) |', end='')
print()

print('DIGITE DOIS NÚMEROS')
num1 = float(input('Primeiro número: '))
num2 = float(input('Segundo número: '))

print('SELECIONE A OPERAÇÃO DE ACORDO COM O MENU')
op = int(input('Digite o número da operação: '))

while op != 5:
  if op == 1:
    r = num1 + num2
  if op == 2:
    r = num1 - num2
  if op == 3:
    if num2 == 0:
      r = 'Não é possível dividir por 0!'
    else:
      r = num1/num2
  if op == 4:
    r = num1*num2
  if num2 != 0:
    print(f'{menu[op-1][0]}: {r}')
  else:
    print(r)
  op = int(input('Digite o número da operação: '))

print('Você saiu!')
```

**9.** Elabore um programa que deve calcular uma integral, na qual a função deve ser implementada no código fonte. O usuário deve fornecer os limites de integração da função considerada. Divida o intervalo em 10000 partições, e faça um somatório, avaliando a função e somando todos os
resultados.

Exemplo: calcule a integral da função $\ \int_{a}^{b} x^2 \,dx$ no intervalo de $0$ a $1$. Para calcular esta integral, deve ser implementada no código fonte e os intervalos devem ser fornecidos pelo usuário.

O programa deve dividir o intervalo em $10000$ partes, e atribuir a $x$ o valor do limite inferior, ou seja $0$.
A função deve ser avaliada para $x = 0$ e depois disso, para cada uma das $10000$ partições feitas. Todas as vezes que $x$ for avaliado, o resultado deve ser armazenado em um somatório. Este somatório é o resultado da integral.

```python
#________________________________________________________________________________________________
# Algoritmo: calculoIntegral
# Propósito: através de intervalos fornecidos pelo usuário, calcular, aproximadamente, a integral através de uma soma de Riemann.
# Retorno: resultado aproximado da integral via soma de Riemann
#________________________________________________________________________________________________

print('Seja f uma função definida em R, diferenciável e contínua em um intervalo [x1,x2] qualquer.')

print(f'{20*"-"}\n{"f(x) = 2x"}\n{20*"-"}')

x1 = float(input('Digite o limite de integração: '))
x2 = float(input('Digite o limite de integração: '))

if x1>x2:
  x1,x2 = x2,x1

n = 10000                #10000 partições
dx = (x2 - x1)/n         #número de retângulos dentro do gráfico da f.
i = 0

somatorio = 0
while i < n:
    x = x1 + i*dx           #a cada rodada, aumenta-se o acréscimo dx
    f = 2*x                 #a cada rodada, avalia-se a função f.
    somatorio += f*dx       #soma das áreas dos retângulos abaixo de f
    i += 1

mensagem = f'Somatório das áreas das {n} partições no intervalo entre {x1} e {x2} é: {somatorio:.1f}'

print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```
