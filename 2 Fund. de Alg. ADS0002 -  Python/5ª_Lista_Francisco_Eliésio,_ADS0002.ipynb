```python
#________________________________________________________________________________________________
# FUNÇÕES PARA VALIDAR OS INPUTS NAS QUESTÕES
#________________________________________________________________________________________________

#Função que verifica se a string do input pode ser convertida em int.
def ni(n):
  while True:
    try:
      n = int(n)
      return n
      break
    except (ValueError, TypeError):
      print('Você deve digitar um número inteiro!')
      n = input('Digite um número: ')
      return ni(n)

#Função que verifica se a string do input pode ser convertida em float.
def nf(n):
  while True:
    try:
      n = float(n)
      return n
      break
    except (ValueError, TypeError):
      print('Você deve digitar um número!')
      n = input('Digite um número: ')
      return nf(n)
```

**7.** Defina uma função recursiva que calcule o maior divisor comum (M.D.C.) entre dois números a e b, em que a > b.

$mdc(a,b)\begin{cases}
a & b = 0\\
mdc(b, a-b|\tfrac{a}{b}|) & a > b\\
\end{cases}$

• Em que $|\tfrac{a}{b}|$ pode ser escrito em Python como: a % b.

```python
#________________________________________________________________________________________________
#Algoritmo: mdc
#Propósito: receber dois números de calcular, via função, o máximo divisor comum entre eles
#Retorno: mdc de dois números
#________________________________________________________________________________________________

def mdc(a,b):
  if b > a:
    a, b = b, a
  if b == 0:
    return a
  else:
    return mdc(b,a % b)

a = ni(input('Digite o valor do primeiro número: '))
b = ni(input('Digite o valor do segundo número: '))

mensagem = f'O mdc entre {a} e {b} é: {mdc(a,b)}.'
print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**8.** Usando a função mdc definida no exercício anterior, defina uma função para calcular o menor múltiplo comum (M.M.C.) entre dois números.

$mmc(a,b)\begin{cases}
\tfrac{|a \times b|}{mdc(a,b)} \\
\end{cases}$

• Em que $|a \times b|$ pode ser escrito em Python como: abs(a*b).

```python
#________________________________________________________________________________________________
#Algoritmo: mmc
#Propósito: retornar o mínimo múltiplo comum entre dois números, calculado via função.
#Retorno: mínimo múltiplo comum
#________________________________________________________________________________________________

def mmc(a,b):
  if b > a:
    a , b = b, a
  def mdc(a,b):
    if b == 0:
      return a
    else:
      return mdc(b,a % b)
  return abs(a*b)/mdc(a,b)

a = ni(input('Digite o valor do primeiro número: '))
b = ni(input('Digite o valor do segundo número: '))

mensagem = f'O mmc de {a} e {b} é: {int(mmc(a,b))}.'
print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**9.** Defina uma função recursiva para calcular a sequência Fibonacci. A sequência em questão começa com dois números 0 e 1. Os números seguintes são a soma dos dois anteriores. A sequência ficaria assim: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...

$fibonacci(n)\begin{cases}
n & n \leq  1\\
fibonacci(n-1)+fibonacci(n-2)\\
\end{cases}$

```python
#________________________________________________________________________________________________
# Algoritmo: sequenciaFibonacci
# Propósito: calcula através de uma função recurvisa n termos da sequência de Finonacci
# Retorno: sequência de Fibonacci
#________________________________________________________________________________________________
def fib(n):
  if n == 0 or n == 1:
    return n
  else:
    return fib(n-1)+fib(n-2)

termos = 0
while termos <= 0:
  termos = int(input('Número de termos para a sequência de Fibonacci: '))

mensagem = ''
for i in range(termos):
  if i < termos -1:
    n = f'{fib(i)}, '
  else:
    n = f'{fib(i)},...'
  mensagem += n

mensagem = f'Sequência: {mensagem}'
print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**10.** Escreva uma função recursiva para calcular a função de Ackerman para dois inteiros positivos m e n, conforme definição.

$A(m,n)\begin{cases}
n+1 & se\ m=0 \\
A(m-1,1) & se\ m>0\ e\ n=0 \\
A(m-1,A(m,n-1)) & se\ m>0\ e\ n>0\\
\end{cases}$

```python
#________________________________________________________________________________________________
# Algoritmo: funcaoArkerman
# Propósito: calcular a função de Arkerman para dois inteiros positivos
# Retorno: retorno da função de Ackerman
#________________________________________________________________________________________________

def A(m,n):
  if m < 0 or n <0:
    return 'Digite inteiros positivos!'
  else:
    if m == 0:
      return n+1
    elif m > 0 and n == 0:
      return A(m-1,1)
    elif m > 0 and n > 0:
      return A(m-1,A(m,n-1))
A(3,6)
```
