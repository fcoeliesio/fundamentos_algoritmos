**2.** Faça um algoritmo que receba dois números e ao final mostre a soma subtração, multiplicação e a divisão dos números lidos.

**3.** Escrever um algoritmo para determinar o consumo médio de um automóvel sendo fornecida a distância total percorrida pelo automóvel e o total gasto de combustível.

```python
consumo  = float(input('Informe a distância percorrida (km): '))/float(input('Informe o consumo de combustível (L): '))

print(f'O consumo médio do automóvel corresponte a {consumo:.2f} km/L')
```

**4.** Escrever um algoritmo que leia o nome de um vendedor, o seu salário fixo, e o total de vendas efetuadas por ele no mês (em dinheiro). Sabendo que este vendedor ganha 15% de comissão sobre suas vendas efetuadas, informar o seu nome, o salário fixo e o salário ao final do mês.

```python
nome = str(input('Vendedor/a - nome: '))
salarioFixo = float(input(f'Salário fixo de {nome} (R$): '))
totalVendas = float(input(f'Total de vendas mensais de {nome} (R$): '))

print(f'Vendedor/a: {nome}\nSalário fixo: R$ {salarioFixo:.2f}\nSalário final: R$ {salarioFixo+0.15*totalVendas:.2f}')
```

**5.** Escrever um algoritmo que leia o nome de um aluno e as notas das três provas que ele obteve no semestre. No final informar o nome do aluno e a sua média (aritmética)

```python
nome = str(input('Estudante - nome: '))

notas = []
print(f'Digite as notas de {nome}:')
for i in range(3):
  notas.append(float(input(f'{i+1}ª nota: ')))

mensagem = f'Estudante: {nome}\nMédia: {(sum(notas)/len(notas)):.2f}'
print(mensagem)
```

**7.** Elaborar um algoritmo que efetue a apresentação do valor da conversão em real (R$$) de um valor lido em dolar (US$). O algoritmo deverá solicitar o valor da cotação do dólar e também a quantidade de dólares disponíveis com o usuário.

```python
cotacaoDolar = float(input('Cotação do dolar (R$): '))
qtdDolar = float(input('Valor em dolares disponível: '))
valorReal = qtdDolar*cotacaoDolar

mensagem = f'O valor convertido em real é igual R$ {valorReal:.2f}.'
print(mensagem)
```

**8.** Ler uma temperatura em graus Celsius e apresentá-la convertida em graus Fahrenheit. A formula de conversão e: $F = \tfrac{9}{5}C+32$, sendo $F$ temperatura em Fahrenheit e $C$ temperatura em Celsius.


```python
C = float(input('Temperatura (°C): '))
while(C < -273.15):
  print('Forneça uma temperatura igual ou superior a -273,15 °C!')
  C = float(input('Temperatura (°C): '))

F = (9/5)*C+32

mensagem = f'Resultado: {C}°C é equivalente a {F:.2f} °F'
print(mensagem)
```

**9.** Faça um algoritmo que receba o preço de custo de um produto e mostre o valor de venda. Sabe-se que o preço de custo recebera um acréscimo de acordo com um percentual informado pelo usuário.

```python
precoCusto = float(input('Preço do custo do produto (R$): '))
percentual = float(input('Acréscimo em valor percentual (0% - 100%): '))
precoFinal = precoCusto*(1+percentual/100)

mensagem = f'O preço de venda é igual á R$ {precoFinal:.2f}.'
print(mensagem)
```
