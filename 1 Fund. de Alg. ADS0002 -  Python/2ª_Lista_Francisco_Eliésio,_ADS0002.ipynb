**1.** Faça um programa que receba três notas de um aluno, calcule e mostre a média aritmética e a mensagem que segue a tabela abaixo. Para alunos de exame, calcule e mostre a nota que devera ser tirada no exame para aprová-lo, considerando que a média no exame e 6,0.

\begin{array} {cc} \hline
MÉDIA \ ARITMÉTICA & MENSAGEM \\ \hline
média < 3 & Reprovado \\ \hline
3 ≥ média < 7 & Exame \\ \hline
média ≥ 7 & Aprovado \\ \hline
\end{array}

```python
def nota(i):
  nota = float(input(f'Digite a {i+1}ª nota: '))
  while(nota < 0 or nota > 10):
    nota = float(input(f'Digite a {i+1}ª nota [0-10]: '))
  return nota

notas = []
for i in range(3): notas.append(nota(i))

media = int((sum(notas)/len(notas))*10)/10

if media >= 3 and media < 7:
  mensagem = f'Necessário nota igual ou maior que {(12 - media):.1f} no Exame Final para aprovação.'
else:
  mensagem = f'Aprovado com média {media:.1f}.' if media >= 7 else f'Reprovado com média {media:.1f}.'

print(mensagem)
```

**5.** Faça um programa que receba o dia o mês e o ano e mostre nos seguintes formatos: dia/mês/ano -> mês por extenso. (ex: 19/03/2020 -> 19 de março de 2020).

```python
meses = {
    'janeiro': 31,
    'fevereiro': 28,
    'março': 31,
    'abril': 30,
    'maio': 31,
    'junho': 30,
    'julho': 31,
    'agosto': 31,
    'setembro': 30,
    'outubro': 31,
    'novembro': 30,
    'dezembro': 31
}

def formatar_data(dia, mes, ano):
    return f'{dia:02}/{mes:02}/{ano:04}'

while True:
    dia = int(input('Digite o dia em formato numérico: '))
    if 1 <= dia <= 31:
        break

while True:
    mes = int(input('Digite o mês em formato numérico: '))
    if 1 <= mes <= 12:
        break

while True:
    ano = int(input('Digite o ano em formato numérico: '))
    if ano > 0:
        break

nomes_mes = list(meses.keys())
mes_nome = nomes_mes[mes - 1]  # Obtém o nome do mês com base no número inserido

# Verificando se o ano é bissexto e ajustando fevereiro, se necessário
if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    meses['fevereiro'] = 29  # Fevereiro tem 29 dias em anos bissextos

# Verificando se o dia é válido para o mês escolhido
if dia > meses[mes_nome]:
    print('Data inválida!')
else:
    # Formatação da data para o formato por extenso
    data_extenso = f'{dia:02} de {mes_nome} de {ano:04}.'
    print(f'Data: {data_extenso}')

    # Formatação da data no formato dd/mm/aaaa
    data_formatada = formatar_data(dia, mes, ano)
    print(f'Data formatada: {data_formatada}')
```

**7.** Faça um programa que receba o salário de um funcionário, calcule e mostre o novo salário desse funcionário, acrescido de bonificação e de auxílio-escola

\begin{array}{|c|c|} \hline
\textbf{SALÁRIO} & \textbf{BONIFICAÇÃO} \\ \hline
\text{Até R\$ 500,00} & 5\%\ \text{do salário} \\ \hline
\text{Entre R\$ 500,00 e R\$ 1.200,00} & 12\%\ \text{do salário} \\ \hline
\text{Acima de R\$ 1.200,00} & \text{Sem bonificação} \\ \hline
\end{array}

\begin{array}{|c|c|} \hline
\textbf{SALÁRIO} & \textbf{AUXÍLIO ESCOLA} \\ \hline
\text{Até R\$ 600,00} & R\$ 150,00\\ \hline
\text{Mais de R\$ 600,00} & R\$ 100,00 \\ \hline
\end{array}






```python
salario = -1
while salario < 0:
  salario = float(input('Salário do funcionário (R$): '))

bonificacao = 0.12*salario if (salario > 500 and salario <= 1200) else 0.05*salario if salario <= 500 else 0
auxilio = 150 if salario <= 600 else 100
salarioTotal = salario + bonificacao + auxilio

print(f'Bonificação (R$): {bonificacao:.2f}')
print(f'Auxílio Escola (R$): {auxilio:.2f}')
print(f'Salário Total (R$): {salarioTotal:.2f}')
```

**8.** Faça um programa para resolver equações de $2º$ grau do tipo $ax^2+bx+c = 0$, em que $\ a, b, c \in \mathbb{R}%$ com $a \neq 0$.

```python
import math

def calcular_delta(a, b, c):
  return b**2 - 4*a*c

def calcular_raizes(a, b, delta):
  if delta < 0:
    return 'A equação não possui raízes reais!'
  elif delta == 0:
    x = -b / (2 * a)
    return f'A equação possui uma única raiz: {x}'
  else:
    x1 = (-b + math.sqrt(delta)) / (2 * a)
    x2 = (-b - math.sqrt(delta)) / (2 * a)
    return f'As raízes da equação são: x1 = {x1}, x2 = {x2}'

a = 0
while a == 0:
  a = float(input('Digite o valor de a: '))
b = float(input('Digite o valor de b: '))
c = float(input('Digite o valor de c: '))

delta = calcular_delta(a, b, c)
resultado = calcular_raizes(a, b, delta)

print(resultado)
```

**9.** Dados três valores $X$, $Y$ e $Z$, verificar se eles podem ser os comprimentos dos lados de um triângulo e, se forem, verificar se o um triângulo equilátero, isósceles ou escaleno. Se eles não formarem um
triângulo escrever uma mensagem. Considerar que:


*   o comprimento de cada lado de um triângulo é menor que a soma dos outros dois lados:
*   chama-se triângulo equilátero o triângulo que tem três lados iguais;
*   chama-se triângulo isósceles o triângulo que tem o comprimento de dois lados iguais;
*   chama-se triângulo escaleno o triângulo que tem os três lados diferentes.

```python
#________________________________________________________________________________________________-
# Algoritmo: condicaoExisteciaTriangulo
# Propósito: Verificar a condição de existência de um triângulo dado os valores dos três lados.
# Retorno: Existência de triângulo e se ele é equilatero, escaleno ou isósceles.
#________________________________________________________________________________________________

print('Condição de existência de um triângulo XYZ.')
x = float(input('Digite o valor do lado X: '))
y = float(input('Digite o valor do lado Y: '))
z = float(input('Digite o valor do lado Z: '))

existeTriangulo = False
lista = ['equilátero','escaleno','isósceles']

if x<(y+z) and y<(x+z) and z<(x+y):
  existeTriangulo = True
  if(x == y and y == z):
    tipo = lista[0]
  elif( x!= y and y != z and x != z):
    tipo = lista[1]
  else:
    tipo = lista[2]
  mensagem = f'O triângulo existe e é {tipo}.'
else:
  mensagem = 'Não existe triângulo com essas medidas.'

print(f'{len(mensagem)*"-"}\n{mensagem}\n{len(mensagem)*"-"}')
```

**10.** Faça um programa que receba:
*   o código de um produto comprado, supondo que a digitação do código do produto seja sempre válida, ou seja, um número inteiro entre $1$ e $10$;
*   o peso do produto em quilos;
*   o código do país de origem, supondo que a digitação do código do país seja sempre válida, ou seja, um número inteiro entre $1$ e $3$.

\begin{array}{|c|c|c|c|} \hline
\textbf{CÓDIGO DO PAÍS DE ORIGEM} & \textbf{IMPOSTO} & \textbf{CÓDIGO DO PRODUTO} & \textbf{PREÇO POR GRAMA} \\ \hline
1 & 0\% & 1 a 4 & 10 \\ \hline
2 & 15\% & 5 a 7 & 25 \\ \hline
3 & 25\% & 8 a 10 & 35 \\ \hline
\end{array}


Calcule e mostre:
*   O peso do produto convertido em gramas.
*   O preço total do produto comprado.
*   O valor do imposto, sabendo-se que ele é cobrado sobre o preço total do produto comprado e depende do país de origem.
*   O valor total, preço total do produto mais imposto.

```python
produto = [[1,10],[2,10],[3,10],[4,10],[5,25],[6,25],[7,25],[8,35],[9,35],[10,35]]
imposto = [[1,0],[2,0.15],[3,0.25]]

codProduto = massa = codPais = 0

while codProduto < 1 or codProduto > 10:
  codProduto = int(input('Próduto - código: '))
while massa <= 0:
  massa = float(input('Massa do produto (kg): '))
while codPais < 1 or codPais > 3:
  codPais = int(input('País de origem - código: '))

massaGrama = massa*1000
preco = massaGrama*produto[codProduto-1][1]
imposto = preco*imposto[codPais-1][1]
valorTotal = preco + imposto

mensagem = f'Massa do produto (g): {massaGrama:.0f}'+'\n'+f'Preço do produto: R$ {preco:.2f}'+'\n'+f'Impostos: R$ {imposto:.2f}'+'\n'+f'Valor total: R$ {valorTotal:.2f}'
print(mensagem)
```
